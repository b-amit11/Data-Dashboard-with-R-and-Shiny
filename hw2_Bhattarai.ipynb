```python
#Normalization with min-max scaling

df_normalized = (df_csv - df_csv.min()) / (df_csv.max() - df_csv.min())

print("Normalized dataframe: ")
print(df_normalized)
```

# 5c) A new column with
# Weighted Total = 0.2(𝑄𝑢𝑖𝑧1) + 0.2(𝑄𝑢𝑖𝑧2) + 0.25(𝑀𝑖𝑑𝑡𝑒𝑟𝑚) + 0.35(𝐹𝑖𝑛𝑎𝑙)

```python
df_csv["Weighted Total"] = (
    0.2 * df_csv["Quiz1"] +
    0.2 * df_csv["Quiz2"] +
    0.25 * df_csv["Midterm"] +
    0.35 * df_csv["Final"]
)

print("New column with weighted total: ")
print(df_csv)
```

# 5d) Ranking students based on weighted total

```python
# Rank students by weighted total
df_csv["Rank"] = df_csv["Weighted Total"].rank(
    ascending=False,
    method="min"
).astype(int)

print(df_csv.sort_values("Rank"))
```

# B) Employees.csv


# Import pandas and read the csv file


```python
import pandas as pd

df = pd.read_csv("employees.csv")

print(df.head())
```

# 1) The summary of the "employees.csv" file  

```python
print(df)

print("\nDataFrame Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

# 2) The missing values and calculate the average salary and bonus


```python
# Fill missing values using mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Bonus %"] = df["Bonus %"].fillna(df["Bonus %"].mean())

# Calculate averages
print("Average Salary:", df["Salary"].mean())
print("Average Bonus:", df["Bonus %"].mean())
```

# 3) The rows to only show salaries more than 100000

```python
high_salary = df[df["Salary"] > 100000]

print("Employees earning more than $100,000:")
print(high_salary)
```

# 4) The male and female employees with maximum and minimum salaries

```python
# Separate male and female employees
male = df[df["Gender"] == "Male"]
female = df[df["Gender"] == "Female"]

# Male employees
print("Male with highest salary:")
print(male.loc[male["Salary"].idxmax(), ["First Name", "Salary"]])

print("\nMale with lowest salary:")
print(male.loc[male["Salary"].idxmin(), ["First Name", "Salary"]])

# Female employees
print("\nFemale with highest salary:")
print(female.loc[female["Salary"].idxmax(), ["First Name", "Salary"]])

print("\nFemale with lowest salary:")
print(female.loc[female["Salary"].idxmin(), ["First Name", "Salary"]])
```

# 5) The minimum and maximum salaries for each team

```python
team_salary = df.groupby("Team")["Salary"].agg(["min", "max"])

print("Minimum and Maximum Salary by Team:")
print(team_salary)
```

# 6) The maximum and average(mean) salary of both genders#

```python
gender_salary = df.groupby("Gender")["Salary"].agg(["max", "mean"])

print("Maximum and Average Salary by Gender:")
print(gender_salary)
```

# 7)The mean salary by team and gender

```python
team_gender = df.groupby(["Team", "Gender"])["Salary"].mean()

print("Average Salary by Team and Gender:")
print(team_gender)
```

# 8) Calculating the total salary by also including bonus

```python
# Calculate total salary including bonus
df["Total Salary"] = df["Salary"] + (
    df["Salary"] * df["Bonus %"] / 100
)

print(df)
```

# 9) Filtering the employees who joined from 2000 to 2010

```python
# Convert Start Date to datetime
df["Start Date"] = pd.to_datetime(df["Start Date"])

# Filter employees from 2000 to 2010
employees_2000_2010 = df[
    df["Start Date"].dt.year.between(2000, 2010)
]

print("Employees who joined between 2000 and 2010:")
print(employees_2000_2010)
```

# 10) Finding the difference between maximum and minimum salry and bonus uing apply()

```python
# Calculate difference using apply()
difference = df[["Salary", "Bonus %"]].apply(max) - df[["Salary", "Bonus %"]].apply(min)

print("Difference between Maximum and Minimum:")
print(difference)
```

C) sales_product.csv

# 1) Reading the "sales_product.csv" using pandas and showing its information

```python
import pandas as pd

df = pd.read_csv("sales_product.csv")

print(df.head())

print("\nDataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nStatistical Summary:")
print(df.describe())
```

# 2) Finding the rows with missing values and removing them

```python
print("Rows with missing values:")
print(df[df.isnull().any(axis=1)])

print("\nMissing Values:")
print(df.isnull().sum())

df = df.dropna()

print("\nAfter removing missing values:")
print(df.isnull().sum())
```

# 3) Changing the columns with appropriate data types

```python
# Remove repeated column headings
df = df[df["Order Date"] != "Order Date"].copy()

# Change data types
df["Order ID"] = pd.to_numeric(df["Order ID"])

df["Quantity Ordered"] = pd.to_numeric(df["Quantity Ordered"])

df["Price Each"] = pd.to_numeric(df["Price Each"])

df["Order Date"] = pd.to_datetime(
    df["Order Date"], format="%m/%d/%y %H:%M"
)

print(df.dtypes)
```

# 4) Creating two new columns month and city by correctly parsing order date and purchase address

```python
# Extract month
df["Month"] = df["Order Date"].dt.month

# Extract city
df["City"] = df["Purchase Address"].str.split(",").str[1].str.strip()

print(df[["Month", "City"]].head())
```

# 5) The month with best sales and how much was earned

```python
# Calculate revenue
df["Revenue"] = df["Price Each"] * df["Quantity Ordered"]

# Total sales for each month
monthly_sales = df.groupby("Month")["Revenue"].sum()

print(monthly_sales)

print("\nBest Month:", monthly_sales.idxmax())
print("Total Sales: $", round(monthly_sales.max(), 2))
```

# 6) The city that sold the most products

```python
city_sales = df.groupby("City")["Quantity Ordered"].sum()

print(city_sales.sort_values(ascending=False))

print("\nCity with most products sold:")
print(city_sales.idxmax())

print("Total Quantity:", city_sales.max())
```

# 7) The product that sold the most per quantities sold

```python
product_sales = df.groupby("Product")["Quantity Ordered"].sum()

print(product_sales.sort_values(ascending=False))

print("\nMost Sold Product:")
print(product_sales.idxmax())

print("Quantity Sold:", product_sales.max())
```

# 8) Parsing the order date properly and extracting the quarter that generates the highest revenue.

```python
df["Month"] = df["Order Date"].dt.month
df["Year"] = df["Order Date"].dt.year
df["Quarter"] = df["Order Date"].dt.quarter

quarter_sales = df.groupby("Quarter")["Revenue"].sum()

print(quarter_sales)

print("\nHighest Revenue Quarter:", quarter_sales.idxmax())
print("Revenue: $", round(quarter_sales.max(), 2))
```

# 9) Computing the revenue and classifying it by top 3 months and best performing city

```python
df["Revenue"] = df["Price Each"] * df["Quantity Ordered"]

# Top 3 months
top_months = df.groupby("Month")["Revenue"].sum()

print("Top 3 Months:")
print(top_months.sort_values(ascending=False).head(3))

# Best city
city_revenue = df.groupby("City")["Revenue"].sum()

print("\nBest Performing City:")
print(city_revenue.sort_values(ascending=False).head(1))
```

# 10) Calculating revenue contribution per product

```python
# Revenue for each product
product_revenue = df.groupby("Product")["Revenue"].sum()

# Total revenue
total_revenue = df["Revenue"].sum()

# Calculate percentage
revenue_percent = (product_revenue / total_revenue) * 100

print(revenue_percent.sort_values(ascending=False))

print("\nProduct with Highest Contribution:")
print(revenue_percent.idxmax())

print("Contribution: ", round(revenue_percent.max(), 2), "%")
```
